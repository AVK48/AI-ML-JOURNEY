EDA (Explorataory Data Analysis):
Exploratory Data Analysis (EDA) is a statistical and programmatic workflow used to systematically analyze, summarize, and transform datasets before applying inferential statistics or training machine learning algorithms.

Tools Used in python:

Pandas: For loading, manipulating, and cleaning data structures.

Matplotlib & Seaborn: For creating visual representations like histograms, scatter plots, and box plots to see the data's distribution.

NumPy: For performing numerical and mathematical operations.

```python
fig, axes = plt.subplots(2,2, figsize = (10,8))

axes[0,0].hist(df['Unit price'],bins =50)
axes[0,0].axvline(df['Unit price'].min(),color = 'black',ls = 'dashed', linewidth = 2, label = "Min")
axes[0,0].set_title("Unit Price distribution")
axes[0,0].set_xlabel("Unit Price")
axes[0,0].set_ylabel("Frequency")
```

-->axvline stands for axis vertical line.

In Matplotlib, it is a function used to draw a straight vertical line across your entire plot at a specific x-coordinate.

--> axhline stands for axis horizontal line

In Matplotlib, it is a function used to drwa a straight horizontal line across your entire plot at a specific y-coordinate


-->linestyle (often abbreviated as ls) is a parameter in Matplotlib used to change the visual pattern of a line in your plot.

Instead of just drawing a standard solid line, you can use linestyle to make the line dashed, dotted, or a combination of both. This is especially helpful in EDA when you have multiple lines on a single chart and need to distinguish between them (like plotting both the mean and the median).
