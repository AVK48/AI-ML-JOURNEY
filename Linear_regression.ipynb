```python
from sklearn.model_selection import train_test_split
X = housing[['age_years','bedrooms','sqft']]
y = housing['price_lakhs']
X_train,X_test, y_train, y_test = train_test_split(X,y,test_size =0.2, random_state = 45)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train)
predictions = model.predict(X_test)
print(predictions)
print(y_test)
```

```python
for feature,coef in zip(X.columns, model.coef_):
  print(f"{feature}:{coef}")
print(f"Intercept:{model.intercept_}")
```

```python
plt.scatter(X_test['sqft'], y_test, color='blue', label='Actual')

plt.plot(X_test['sqft'], predictions, color='red', label='Regression Line')


plt.xlabel('Square Footage (sqft):')
plt.ylabel('House Price (In Lakhs):')
plt.title('Simple Linear Regression: Square Footage vs. House Price')
plt.legend()
plt.show()
```
