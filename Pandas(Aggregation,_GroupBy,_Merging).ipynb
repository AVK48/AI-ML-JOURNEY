GROUPBY

```python
import pandas as pd
data = {
    "name":       ["Alice", "Bob", "Charlie", "Diana", "Eve", "Frank"],
    "department": ["HR", "Tech", "Tech", "HR", "Finance", "Tech"],
    "salary":     [52000, 85000, 91000, 48000, 73000, 88000],
    "rating":     [4, 5, 5, 3, 4, 4]
}
df = pd.DataFrame(data)
```

```python
df.groupby("department")["salary"].mean()
```

```python
df.groupby("department")["rating"].mean()
```

```python
df.groupby("department")["rating"].count()
```

```python
df.groupby("department")["salary"].count()
```

```python
df.groupby("department")["salary"].max()
```

```python
df.groupby("department")["salary"].min().index
```

```python
df["department"].value_counts(normalize=True)
```

```python
summary =df.groupby("department").aggregate(headcount=("name","count"),
                                            avg_salary = ("salary","mean"),
                                            max_salary= ("salary","max"),
                                            avg_rating =("rating","mean"))
summary
```

```python
df["department"].max()
```

```python
idx =df.groupby("department")["salary"].idxmax()
idx
```

```python
top_earners=df.loc[idx]
top_earners
```
