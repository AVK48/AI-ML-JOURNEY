```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#plot
x= [1,2,3,4]
y=[5,6,7,8]
plt.plot(x,y) #creating a plot for data
plt.grid() #visualizing the grid in the plots
plt.show() #showing the plot
```

```python
#pyplot API
#Univariate plot - is a chart or graph used to analyse and visualize the data of a single variable
data = {
    "Salary" : [1200,13300,2345,2357,9675,3254,6548,5478]
}
df = pd.DataFrame(data)
df.head()
```

```python
df.shape
```

```python
#line plot
plt.plot(df['Salary'],color = 'red', marker = 'o', linestyle= ':', linewidth = '2') # ploting the salary data, with color = red, marking = o, linestyle = :, linwidth = 2
plt.grid() #ploting the grids
plt.show()# showing the plot
```

```python
#Histogram - A histogram is a specific type of univariate graph used to visualize the frequency distribution of continuous numerical data
plt.hist(df['Salary'],bins = 6, color= 'yellow') # creating a histogram for salary with 5 parts called bins using color yellow
plt.show() #show the histogram
```

```python
#Boxplot -A boxplot (also known as a box-and-whisker plot) is a standardized way of displaying the distribution of a dataset based on a five-number statistical summary.
plt.boxplot(df['Salary'])
plt.show()
```

```python
#univariate categorical
```

```python
df['dept'] = ['hr','it','finance','it'] * 2
```

```python
df.shape #to know the dimension count of the data frame
```

```python
df.info() #to know the information of the dataframe
```

```python
count = df['dept'].value_counts() #to get the unique departments in the dept column
count
```

```python
#pie chart- a pie chart is a circular stastical graph divided into slices to illustrate the data. it is a type fo univariate plot specially used for categorical data to show the differences
plt.pie(count, labels= count.index, autopct = '%1.3f', explode = [0,0.1,0.2]) # creating pie chart, by labeling the indexes, values of data, sepearte the data using explode
plt.axis('equal') #used to maintain the circle
plt.show()
```

```python
#countplot - it is type of bar chart used to display the number of occurrences for each category in a categorical data set
plt.bar(count.index, count, color = ['yellow','green','red'])
plt.show()
```

```python
df['age'] = [21,41,42,36,26,48,24,26]
```

```python
# bivariate - Bivariate is a chart or graph used to analyze and visualize the multi variables of the data
#scatterplot - A scatter plot is a classic bivariate graph used to visualize the relationship (or correlation) between two distinct numerical variables.
plt.scatter(df['age'],df['Salary'],color = 'blue')
plt.grid()
plt.show()
```

```python
salary_by_dept = df.groupby('dept')['Salary'].sum()
salary_by_dept
plt.pie(salary_by_dept, labels = salary_by_dept.index, autopct = '%1.2f')
plt.show()
```
