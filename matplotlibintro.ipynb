```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
sales_df = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "online":  [120, 135, 150, 142, 160, 175],
    "store":   [80,  90,  95,  88,  100, 110],
    "region":  ["North", "North", "South", "South", "North", "South"]
})
sales_df
```

```python
plt.plot(sales_df['online'],sales_df['region'])
plt.show()
```

```python
plt.plot(sales_df['month'],sales_df['online'])
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.plot(sales_df['month'],sales_df['store'],sales_df['online'])
plt.ylabel("Store Sales & Online Sales")
plt.xlabel("Month")
plt.title("Sales over the time")

plt.legend(["store","online"])
plt.show()
```

marking symbols(joining symbols)

```python
plt.figure(figsize=(10,5))
plt.plot(sales_df['month'],sales_df['store'],marker='o')
plt.plot(sales_df['month'],sales_df['online'],marker='*')
plt.ylabel("Store Sales & Online Sales")
plt.xlabel("Month")
plt.title("Sales over the time")
plt.legend(["store","online"])
plt.show()
```

title fontsize & grid

```python
plt.figure(figsize=(10,5))
plt.plot(sales_df['month'],sales_df['store'],marker='o')
plt.plot(sales_df['month'],sales_df['online'],marker='*')
plt.ylabel("Store Sales & Online Sales")
plt.xlabel("Month")
plt.title("Sales over the time",fontsize=10)
plt.grid(axis='y',alpha=1)
plt.legend(["store","online"])
plt.tight_layout()
plt.show()
```
